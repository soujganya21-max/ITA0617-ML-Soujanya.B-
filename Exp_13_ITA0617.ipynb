```python

from sklearn.linear_model import LinearRegression

X = [[1], [2], [3], [4]]
y = [100000, 200000, 300000, 400000]

m = LinearRegression()
m.fit(X, y)

print("Predicted Car Price:", m.predict([[5]]))
```
