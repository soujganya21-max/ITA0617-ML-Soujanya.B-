```python

from sklearn.linear_model import LinearRegression

X = [[1], [2], [3], [4], [5]]
y = [2, 4, 6, 8, 10]

m = LinearRegression()
m.fit(X, y)

print("Prediction:", m.predict([[6]]))
```
