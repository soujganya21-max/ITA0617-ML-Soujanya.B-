```python

from sklearn.neural_network import MLPClassifier

X = [[0, 0], [0, 1], [1, 0], [1, 1]]
y = [0, 1, 1, 0]

ann = MLPClassifier(hidden_layer_sizes=(2,), max_iter=1000, random_state=1)

ann.fit(X, y)

print("Prediction:", ann.predict([[1, 0]]))
```
