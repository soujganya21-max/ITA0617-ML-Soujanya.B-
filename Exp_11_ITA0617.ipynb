```python

from sklearn.tree import DecisionTreeClassifier

X = [[500], [600], [700], [800]]
y = ['Poor', 'Fair', 'Good', 'Excellent']

m = DecisionTreeClassifier(random_state=0)
m.fit(X, y)

print("Prediction:", m.predict([[750]]))
```
