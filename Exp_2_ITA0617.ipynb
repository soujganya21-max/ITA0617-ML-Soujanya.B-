```python

data = [
    ['Sunny', 'Warm', 'Yes'],
    ['Sunny', 'Cold', 'Yes'],
    ['Rainy', 'Warm', 'No'],
    ['Sunny', 'Warm', 'Yes']
]

S = data[0][:-1]
G = ['?'] * len(S)

for row in data[1:]:
    if row[-1] == 'Yes':
        for i in range(len(S)):
            if S[i] != row[i]:
                S[i] = '?'
    else:
        for i in range(len(G)):
            if S[i] != row[i]:
                G[i] = S[i]

    print("S =", S, "G =", G)

print("Final S =", S)
print("Final G =", G)
```
