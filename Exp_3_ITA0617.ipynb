```python

from sklearn.tree import DecisionTreeClassifier

X = [[0, 0], [0, 1], [1, 0], [1, 1]]
y = ['No', 'No', 'Yes', 'Yes']

model = DecisionTreeClassifier(criterion='entropy')
model.fit(X, y)

result = model.predict([[1, 0]])

print("Predicted Result:", result[0])
```
