```python

from sklearn.linear_model import LogisticRegression

X = [[1], [2], [3], [6], [7], [8]]
y = [0, 0, 0, 1, 1, 1]

m = LogisticRegression()
m.fit(X, y)

print("Prediction:", m.predict([[5]]))
```
