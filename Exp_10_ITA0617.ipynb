```python

import numpy as np
from sklearn.mixture import GaussianMixture

X = np.array([[1], [2], [3], [10], [11], [12]])

g = GaussianMixture(n_components=2, random_state=0)
g.fit(X)

print("Clusters:", g.predict(X))
```
